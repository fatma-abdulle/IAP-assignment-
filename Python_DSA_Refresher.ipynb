### Question 1

```python
# Convert days entered by the user into seconds
days = int(input("Enter the number of days: "))
seconds = days * 24 * 60 * 60
print(f"{days} day(s) is equal to {seconds} seconds.")
```

### Question 2

```python
import math

# Compute the volume of a sphere using (4/3)*pi*r^3
radius = float(input("Enter the radius of the sphere: "))
volume = (4 / 3) * math.pi * (radius ** 3)  # ** is the exponential operator
print(f"The volume of a sphere with radius {radius} is {volume:.2f} cubic units.")
```

### Question 3

```python
# Functions to compute area and perimeter of a square
def calculate_area(side):
    return side * side

def calculate_perimeter(side):
    return 4 * side

side_length = float(input("Enter the side length of the square: "))
area = calculate_area(side_length)
perimeter = calculate_perimeter(side_length)

print(f"Area of the square: {area}")
print(f"Perimeter of the square: {perimeter}")
```

### Question 4

```python
# Function to check whether a character is uppercase or lowercase
def check_case(character):
    if character.isupper():
        print(f"'{character}' is an UPPERCASE letter.")
    elif character.islower():
        print(f"'{character}' is a lowercase letter.")
    else:
        print(f"'{character}' is not a letter.")

user_char = input("Enter a character: ")
check_case(user_char)
```

### Question 5

```python
# REPEAT...UNTIL emulated with while True + break (runs at least once)
x = 0
y = 20

while True:
    y = y - 4
    x = x + 2 / y

    if y < 6:  # UNTIL condition
        break

print("x =", x)
```

### Question 6

```python
# Collect 5 user-input values into a list, then find their average
values = []

for i in range(5):
    value = float(input(f"Enter value {i + 1}: "))
    values.append(value)

average = sum(values) / len(values)

print("Values entered:", values)
print(f"Average of the values: {average:.2f}")
```
